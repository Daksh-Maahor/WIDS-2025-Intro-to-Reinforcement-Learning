```python
import numpy as np
```

```python
def toggle(state, idx, n):
    state ^= (1 << idx)
    r, c = divmod(idx, n)
    for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
        rr, cc = r + dr, c + dc
        if 0 <= rr < n and 0 <= cc < n:
            state ^= (1 << (rr*n + cc))
    return state
```

```python
def solve_lights_out(n):
    num_states = 2 ** (n*n)
    goal = 0

    V = np.full(num_states, 1000.0)
    V[goal] = 0

    policy = np.zeros(num_states, dtype=int)

    while True:
        delta = 0
        for s in range(num_states):
            if s == goal:
                continue

            costs = []
            for a in range(n*n):
                s_next = toggle(s, a, n)
                costs.append(1 + V[s_next])

            best = min(costs)
            delta = max(delta, abs(V[s] - best))
            V[s] = best
            policy[s] = np.argmin(costs)

        if delta < 1e-6:
            break

    return V, policy
```

```python
V, policy = solve_lights_out(3)

print("Maximum minimum moves needed:", int(np.max(V)))
```
